# Injury Data Preprocessing

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

np.set_printoptions(threshold=np.inf, linewidth=np.inf)
```

```python
#

## 
injury_data_path = Path('injury data/')
injury_data_list = []
for file_path in injury_data_path.glob("*.xlsx"):
    df = pd.read_excel(file_path)
    injury_data_list.append(df)
injury_data = pd.concat(injury_data_list, ignore_index=True)

## 
injury_data = injury_data[~injury_data['Pos'].isna()].reset_index(drop=True)

## 
pitcher_pos = ['RP', 'SP', 'RP/SP', 'SP/RP',]
injury_data = injury_data[injury_data['Pos'].isin(pitcher_pos)].reset_index(drop=True)
injury_data = injury_data[~injury_data['NameASCII'].isin(['Sergio Alcantara', 'Daz Cameron', 'Isaac Paredes'])].reset_index(drop=True)

## 
injury_data['Injury / Surgery Date'] = injury_data['Injury / Surgery Date'].apply(lambda x: '06/05/24' if x == "Jun '24" else x)
injury_data['Injury / Surgery Date'] = pd.to_datetime(injury_data['Injury / Surgery Date'], format='%m/%d/%y', errors='coerce')
injury_data['IL Retro Date'] = pd.to_datetime(injury_data['IL Retro Date'], format='%m/%d/%y', errors='coerce')
injury_data['Eligible to Return'] = pd.to_datetime(injury_data['Eligible to Return'], format='%m/%d/%y', errors='coerce')
injury_data['Return Date'] = pd.to_datetime(injury_data['Return Date'], format='%m/%d/%y', errors='coerce')

injury_data = injury_data.drop(columns=['Name'])

injury_data['MLBAMID'] = injury_data['MLBAMID'].astype(int)
```
